```python
import os

import matplotlib.pyplot as plt
import pandas as pd
from sqlalchemy import create_engine

# 1. Conexión a la base de datos PostgreSQL local (Docker)
database_url = os.getenv("DEEPVITAL_DATABASE_URL")
if not database_url:
    raise RuntimeError(
        "DEEPVITAL_DATABASE_URL is not configured. Copy .env.example to .env "
        "and provide a local database URL."
    )
engine = create_engine(database_url)

# 2. Consulta SQL: Extraer una serie temporal
# Nota: Reemplaza 'ID_DEL_PACIENTE' con un ID real que veas en tu tabla de pacientes
query_sensores = """
SELECT fecha_hora, valor, codigo_sensor
FROM signos_vitales 
WHERE paciente_id = 'ID_DEL_PACIENTE' 
ORDER BY fecha_hora ASC
"""

df_sensores = pd.read_sql(query_sensores, engine)
df_sensores['fecha_hora'] = pd.to_datetime(df_sensores['fecha_hora'])
df_sensores.set_index('fecha_hora', inplace=True)

# 3. Filtrar una señal específica (por ejemplo, Frecuencia Cardíaca)
# (El código exacto dependerá del sistema de codificación que extrajimos del FHIR)
frecuencia_cardiaca = df_sensores[df_sensores['codigo_sensor'] == 'codigo_hr_aqui']

# 4. Graficar la serie temporal
plt.figure(figsize=(14, 6))
plt.plot(frecuencia_cardiaca.index, frecuencia_cardiaca['valor'], marker='.', linestyle='-', color='#d62728', linewidth=1)
plt.title('Evolución Temporal de Constantes Vitales en UCI')
plt.xlabel('Tiempo')
plt.ylabel('Valor del Sensor')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
